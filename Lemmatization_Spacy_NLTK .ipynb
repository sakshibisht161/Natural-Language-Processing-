```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

# Basic lemmatization

```python
text = "Programmers create custom classes easily."
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.lemma_)
```

# Lemmatizing Plural Nouns

```python
text = "Directories store hidden files safely."
doc = nlp(text)
for token in doc:
    print(token.text ,"->",token.lemma_)
```

# Lemmatizating verbs

```python
text = "writes writing wrote written"
doc = nlp(text)
for token in doc:
    print(token.text ,"->",token.lemma_)
```

# Lemmatizing different verb forms

```python
text ="I am writing. She wrote notes. They have written books."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text, "->", token.lemma_)
```

# Lemmatization with a sentence

```python
text = "They have driven cars."
doc = nlp(text)
lemmatized_text = " ".join(token.lemma_ for token in doc)
print(lemmatized_text)
```

#  Lemmatization of Irregular Words

# Lemmatization Baed on Part of Speech

```python
text ="newer newest wrote written bright data"
doc = nlp(text)
for token in doc:
  print(
      token.text,
      "POS:", token.pos_,
      "Lemma:", token.lemma_
  )
```

#  Remove Punctuation and Lemmatize

```python
text ="Wow! The children went to the park and saw three mice yesterday."
doc = nlp(text)
words = []
for token in doc:
  if not token.is_punct:
    words.append(token.lemma_)
print(words)
```

#  Lemmatization with Stopword Removal

```python
text = "The quick brown foxes are jumping over the lazy dogs right now"
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)
print(words)
```

#  Create a Lemmatization Function

```python
def lemmatize_text(text):
  doc = nlp(text)

  result = []

  for token in doc :
    if not token.is_stop and not token.is_punct:
      result.append(token.lemma_)
  return " ".join(result)

text = "Many students are writing their final exams in the big library"
result = lemmatize_text(text)
print(result)
```

# Lemmatization using NLTK

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

# Basic NLTK Lemmatization

```python
words =["men", "women", "children", "teeth", "mice"]
for word in words:
  print(word, "->",lemmatizer.lemmatize(word))
```

#  Limmatizing Verbs

```python
words =  ["running", "ate", "swimming", "flies", "went"]
for word in words:
  print(word, "->",lemmatizer.lemmatize(word,pos="v"))
```

# Limmatizing noun

```python
words = ["heroes", "thieves", "babies", "wives", "wolves"]
for word in words:
  print(word, "->",lemmatizer.lemmatize(word,pos="n"))
```

# Limmatizing Adjectives

```python
words =["oxen", "geese", "feet", "phenomena", "crises"]
for word in words:
  print(word, "->",lemmatizer.lemmatize(word,pos="a"))
```

# Limmatizing Adverbs

```python
words = ["fast", "loudly", "quietly", "angrily", "gently"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="r"))
```

# 6 Irregular Verb Lemmatization

```python
words = ["ran", "running", "swam", "swimming", "swum"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="v"))
```

#  Comapare Noun and Verb Lemmatization

```python
word = "building"
print("As noun:",lemmatizer.lemmatize(word,pos="n"))
print("As verb:",lemmatizer.lemmatize(word,pos="v"))
```

# Lemmatizing a List of Words

```python
words = ["leaves", "wolves", "crying", "studying", "better"]
for word in words:
  if word.endswith("ing") or word.endswith("ed"):
    print(word, "->",lemmatizer.lemmatize(word,pos="v"))

  else:
    lemma = lemmatizer.lemmatize(word,pos="n")

  print(word, "->", lemma)
```

# Lemmatization with Tokenization

```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

text = "The birds are flying high."
tokens = word_tokenize(text)
for token in tokens:
 lemma = lemmatizer.lemmatize(token,pos="v")
 print(token, "->", lemma)
```
