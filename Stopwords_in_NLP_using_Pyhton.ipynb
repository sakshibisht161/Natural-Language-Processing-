# Stopwords using spacy

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

#  Check Whether a Word is a Stopword

```python
words = ["running", "cat", "delicious", "not", "always", "computer"]
for word in words:words = ["running", "cat", "delicious", "not", "always", "computer"]
for word in words:
  doc = nlp(word)
for token in doc:
  print(token.text, "->" , token.is_stop)
```

# Identify Stopwords in a sentences

```python
text = "We should learn how to code daily."
doc = nlp(text)
for token in doc:
    print(token.text, "->" , token.is_stop)
```

# Extract Only Stopwords

```python
text = "The quick brown fox jumps over the lazy dog and runs away."
doc = nlp(text)
stopwords = []
for token in doc :
  if token.is_stop:
    stopwords.append(token.text)
print(stopwords)
```

#  Extract Non - Stopwords

```python
text = "Wow! Machine learning is very interesting."
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print(words)
```

# Remove stopwords from a sentence

```python
text = "The developers are building a massive software platform inside the office."
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop:
    result.append(token.text)
print(" ".join(result))
```

# Remove stopword and punctuation

```python
text = "Alice is analyzing the financial dataset, but Bob is writing the final report!"
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text)
print(" ".join(result))
```

# Convert text to lowercase and remove stopwords

```python
text = "Natural Language Processing (NLP) Is Changing The Entire Tech World."
doc = nlp(text)
result = []
for token in doc :
  if not token.is_stop and not token.is_punct:
    result.append(token.text.lower())
print(" ".join(result))
```

#  Create a stopword removal function

```python
def remove_stopwords(text):
  doc = nlp(text)
  result = []
  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.text)
  return " ".join(result)
text = "The engineers are deployment of a new Artificial Intelligence model."
print(remove_stopwords(text))
```

```python
from nltk.sem.drt import Tokens
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
import string
```

# Display NLTK Stopwords

```python
stop_words = set(stopwords.words("english"))
print(stop_words)
print(len(stop_words))
```

# Check whether a word is a stopword

```python
words = ["running", "cat", "delicious", "not", "always", "computer"]
for word in words:
    print(word, "->", word.lower() in stop_words)
```

# Remove stopwords from a sentences

```python
text ="An engineer is working on an advanced software architecture."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(result)
```

# Remove stopwords and punctuation

```python
text ="Hey! The developers are deploying the code, but it has bugs."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)
print(result)
```

# Convert Text to Lowercase and Remove Stopwords

```python
text = "Artificial Intelligence Is Transforming The Modern Industry."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())
print(result)
```

# Count stopwords in a Text

```python
text =  "We should focus on our goals because time is very precious for us."
tokens = word_tokenize(text)
count = 0
for word in tokens:
  if word.lower() in stop_words:
    count += 1
print("Number of stopwords:", count)
```

# Add Custom Stopwords in NLTK

```python
stop_words = set(stopwords.words("english"))
custom_words = {"cloud", "software", "system"}
stop_words.update(custom_words)
text = "The engineer is building a cloud software system."

tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(result)
```
