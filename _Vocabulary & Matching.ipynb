# Vocabulary & Matching

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)
```

# Creating Patterns

```python
pattern1 = [{'LOWER': 'artificialintelligence'}]
pattern2 = [{'LOWER': 'artificial'}, {'LOWER': 'intelligence'}]
pattern3 = [{'LOWER': 'artificial'}, {'IS_PUNCT': True}, {'LOWER': 'intelligence'}]


matcher.add('AI_Match',[pattern1,pattern2,pattern3])

doc = nlp(u"Modern artificialintelligence tools are evolving. "
          u"We love Artificial Intelligence, and even artificial-intelligence "
          u"bots are getting smarter.")
found_matches = matcher(doc)
print(found_matches)
```

```python
for match_id,start,end in found_matches:
    string_id = nlp.vocab.strings[match_id]
    span = doc[start:end]
    print(match_id,string_id,start,end,span.text)
```

# First create a list of match phrases

```python
from spacy.matcher import PhraseMatcher

# 1. Initialize the PhraseMatcher (if not already done)
matcher = PhraseMatcher(nlp.vocab)

# 2. Define your phrases and build doc patterns
phrase_list =[ 'global economy', 
    'automation tools', 
    'generative AI', 
    'technological revolutions']
phrase_patterns = [nlp(text) for text in phrase_list]

# 3. Add to the matcher using the correct spaCy v3+ syntax
matcher.add('HealthcarePatterns', phrase_patterns)

# 4. Find matches in your document
matches = matcher(doc3)
matches
```
