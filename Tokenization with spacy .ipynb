```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

# Basic Tokenizaton

```python
text = "Big data is heavy to load"
doc = nlp(text)
for token in doc :
    print(token.text)
```

# Tokenization with Token Index

```python
text = "Graphs are used to show trends."
doc = nlp(text)
for token in doc :
    print(token.i ,token.text)
```

# Toenization with token attributes

```python
text ="Servers are used to host models."
doc = nlp(text)
for token in doc:
  print(token.text,
        token.is_alpha,
        token.is_punct,
        token.is_space
        )
```

# Tokenization of a sentence with punctuation

```python
text = "Hello! NLTK is superb for tokens."
doc = nlp(text)
for token in doc:
  print(token.text)
```

# Tokenization of numbers

```python
text = "10 gigabytes is heavy to load."
doc = nlp(text)
for token in doc:
  print(token.text)
```

#  Tokenization of Email Addresses

```python
text = "User1@company.com is easy to reach."
doc = nlp(text)
for token in doc:
  print(token.text)
```

# Tokenization of Contractions

```python
text = "Libraries shouldn't be overlooked by developers."
doc = nlp(text)
for token in doc :
    print(token.text)
```

# Tokenization of Hyphenated words

```python
text = "State-of-the-art tech is easy to launch. "
doc = nlp(text)
for token in doc :
    print(token.text)
```

#  Removing Punctuation After Tokenization

```python
text = "Raw-data, though messy, is vital to clean!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

# Getting Only Alphabetic Tokens
# We can use is_alpha to select only alphabetic tokens.

```python
text = "Run 99 is perfect for training."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text)
```

# Converting Tokens into a list

```python
text = "PowerBI/Tableau tool is superior for business."
doc = nlp(text)
tokens = [token.text for token in doc]
print(tokens)
```

# Tokenization of Multiple Sentences

```python
text = "Git-versioning is essential to master. GitHub/GitLab repositories are vital for teams."
doc = nlp(text)
for sent in doc.sents:
  print(sent.text)
```

# Tokenization and Sentences Segmentation
# Spacy can also identify individual sentences .

```python
text = "Cloud-buckets, though remote, are safe to secure.AWS (S3 storage) is ideal for backups."
doc = nlp(text)
for sent in doc.sents:
  print("Sentence:",sent.text)

  for token in sent:
    print("Token:",token.text)
    print()
```

# Tokenization from a Text File

```python
with open("demo.txt","r",encoding ="utf-8") as f:
  text = f.read()
doc = nlp(text)
for token in doc:
  print(token.text)
```
