# Stemming using NLTK 

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

# Basic Stemming

```python
words = ["trainers", "training", "models", "quickly", "algorithms"]
for word in words:
  print(word, "->", stemmer.stem(word))
```

#  Stemming a structure

```python
text = "Trainers are training models quickly."
words = text.split()
for word in words:
  print(word, "->", stemmer.stem(word))
```

# Words ending with "ing"

```python
words = ["executing","cleaning","scaling","gathering","indexing"]
for word in words:
  print(word, "->", stemmer.stem(word))
```

# Words ending with "ed"

```python
words = ["collected","proved","matched","triggered","mapped"]
for word in words:
    print(word ,"->" , stemmer.stem(word))
```

# words ending with "S"

```python
words = ["steps","queues","consumers","scores","workers"]
for word in words:
    print(word , "->" , stemmer.stem(word))
```

# Different forms of the same word

```python
words =["analyze", "analyzes", "analyzed", "analyzing", "analyst", "analytics"]
for word in words :
    print(word , "->" , stemmer.stem(word))
```

# Stemming with a loop

```python
words = ["predict", "predicts", "predicted", "predicting", "predictor", "prediction"]
for word in words:
  stem = stemmer.stem(word)
  print(f"{word:12} -> {stem}")
```

# stemming text using regular expressions

```python
import re
text = "Cleaners are cleaning datasets thoroughly. Systems removed duplication for developers."
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
for word in words:
  print(word, "->", stemmer.stem(word))
```

# create a stemmed sentence

```python
text = "Indexers are indexing documents rapidly. Queries retrieved answers for users."
words = text.lower().split()
stemmed_words = []
for word in words:
  stemmed_words.append(stemmer.stem(word))
stemmed_text = " ".join(stemmed_words)
print("Original Text:", text)
print("Stemmed Text:", stemmed_text)

print("\nstemmed:")
print(stemmed_text)
```

```python
import spacy
from nltk.stem import PorterStemmer
```

```python
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
```

```python
text = "Deployers are deploying applications safely."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->", stemmer.stem(token.text))
```

# Schedulers ,scheduling

```python
text =  "Schedulers are scheduling tasks automatically."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

#  Reaind -related words

```python
text = "A writer is easy to track. The code is written for systems."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

# Stemming a complete paragragh

```python
text = """A reader is quick to read. 
The book is readable for students. 
Metrics are used to measure readability.
"""
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15}, ->, {stemmer.stem(token.text)}")
```

# Removing punctuation before stemming

```python
text = "A SQL-table is quick to query. Inner/outer joins are standard for relational data."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15}, ->, {stemmer.stem(token.text)}")
```

# Stemming with lowercase conversion

```python
text = "Server Monitoring is Automated to Scale."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    word = token.text.lower()
    print(word, "->", stemmer.stem(word))
```

# Create a stemmed sentence using spacy

```python
text = "Loops help us to repeat tasks quickly—saving a lot of development time."
doc = nlp(text)
stemmed_words = []
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))
stemmed_text = " ".join(stemmed_words)
print("Original Text:")
print(text)
print("\nStemmed:")
print(stemmed_text)
```
