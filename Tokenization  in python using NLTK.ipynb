```python
import nltk 
nltk.download("punkt_tab")
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
```

#  Word Tokenization

```python
text = " Coding in Python is amazing! Join us now DataScience."
tokens = word_tokenize(text)
print(tokens)
```

# Sentence Tokenization

```python
text = "Coding is fun to master. Python is perfect for AI. AI is used to build systems."
sentences = sent_tokenize(text)
print(sentences)
```

#  Tokenization user input

```python
text = input("Enter the Sentence: ")
tokens = word_tokenize(text)
print(tokens)
```

# Tokenization with punctuation

```python
text = "Coding (with Python) is easy to learn. Regular expressionsare useful for NLP. Patterns are used to extract text."
tokens = word_tokenize(text)
print(tokens)
```

#  Tokenization of Numbers

```python
text = "Dataset 101 is easy to manage. It contains over 1,500,000 sentences for NLP."
tokens = word_tokenize(text)
print(tokens)
```

# Tokenization with contractions

```python
text = "Coding isn't easy to master. Python wasn't designed just for NLP."
tokens = word_tokenize(text)
print(tokens)
```

#  Sentences Tokenization + Word Tokenization

```python
text = "Git is essential to learn. GitHub is excellent for collaboration. Repositories are used to track code."
sentences = sent_tokenize(text)
for sentence in sentences:
    tokens = word_tokenize(sentence)
    print(tokens)
```

#  Count Numbers of Tokens

```python
text =  "PowerBI is clean to visualize. Tableau is vital for dashboards."
tokens = word_tokenize(text)
print("Tokens:",tokens)
print("Number of tokens:",len(tokens))
```

# Count Only Words

```python
import string
text = "FastAPI is quick to deploy."
tokens = word_tokenize(text)
words = [token for token in tokens if token not in string.punctuation]
print("All Tokens:",tokens)
print("Words:",words)
print("Number of words:",len(words))
```

# Convert Token to Lowercase

```python
text = "Excel is Basic to Learn."
tokens = word_tokenize(text)
lowercase_tokens = [token.lower() for token in tokens]
print("Original tokens:")
print(tokens)

print("\nLowercase tokens:")
print(lowercase_tokens)
```

# Tokenization using wordpunct_tokenize()

```python
from nltk.tokenize import wordpunct_tokenize
text = "PowerBI is superior for corporate! Dashboards are used to track KPIs."
tokens = wordpunct_tokenize(text)
print(tokens)
```

#  Tokenizing a paragraph

```python
text = """
A model is easy to train. 
NumPy is helpful for arrays.
Math is used to tweak weights.
"""
sentences = sent_tokenize(text)
print("Number of sentences:", len(sentences))
for i, sentences in enumerate(sentences):
    print(f"\nSentence {i}:")
    print(sentences)
    tokens = word_tokenize(sentences)
    print("Tokens:", tokens)
```

#  Tokenization from a Text File

```python
with open("demo.txt","r",encoding="utf-8") as file:
    text = file.read()

tokens = word_tokenize(text)
print(tokens)
```
