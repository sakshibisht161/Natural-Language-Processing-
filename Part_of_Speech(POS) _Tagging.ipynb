```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

# Basic Pos Tagging

```python
text = "Machine learning is amazing."
doc = nlp(text)
for token in doc :
  print(token.text , token.pos_)
```

# Display word Pos and explanation

```python
text = "The Developers writes python code."
doc = nlp(text)
for token in doc :
  print(token.text , token.pos_ , spacy.explain(token.pos_))
```

#  Nouns and Verbs

```python
text = "Developers create software using Python at Google."
doc = nlp(text)
for token in doc :
  if token.pos_ in ["NOUN","VERB"]:
    print(token.text , token.pos_)
```

#  Extract Adjective

```python
text = "The quick brown fox jumps over a lazy dog."
doc = nlp(text)
for token in doc :
  if token.pos_ == "ADJ":
    print(token.text , token.pos_)
```

# Extract Adverbs

```python
text = "The team worked incredibly hard and successfully finished the project yesterday."
doc = nlp(text)
for token in doc :
  if token.pos_ == "ADV":
    print(token.text , token.pos_)
```

# Extract ProperNoun

```python
text = "Alice visited Paris to study Art and History at Sorbonne University."
doc = nlp(text)
for token in doc :
  if token.pos_ == "PROPN":
    print(token.text , token.pos_)
```

#  POS Tagging With Numbers

```python
text = "The quick brown fox jumps over 3 lazy dogs."
doc = nlp(text)
for token in doc :
  print(token.text , token.pos_)
```

#  POS Frequency
# We can counthoe many times each POS category occurs

```python
from collections import Counter
text = """
JavaScript is widely used for web applications.
Developers deploy services on cloud platforms like AWS.
Cloud technology scale modern businesses efficiently.
"""
doc = nlp(text)
pos_counts = Counter([token.pos_ for token in doc])
print(pos_counts)
```

#  POS Tagging in a sentence

```python
text = "A clever engineer built a working robot yesterday."
doc = nlp(text)
for token in doc :
  print(f"{token.text:12} , {token.pos_}")
```

# POS Tagging with tag 

```python
text = "He is studying Machine Learning."
doc = nlp(text)
for token in doc :
  print(token.text , token.pos_ , token.tag_)
```

#  POS Tagging with Visualization

```python
from spacy import displacy
text ="The young programmer faced a daunting challenge, but she completed the project with ease."
doc = nlp(text)
spacy.displacy.render(doc , style = "dep", jupyter = True)
```

# Visualize POS Tags as a Bar Chart
# We can count POS tags and display their frequency .

```python
import spacy
from collections import Counter
nlp = spacy.load("en_core_web_sm")
import matplotlib.pyplot as plt
text = """
"Artificial intelligence is transforming modern industries.
Engineers create innovative tools to solve complex problems. 
Creative thinking drives major breakthroughs in technology.
"""
doc = nlp(text)
pos_counts = Counter([token.pos_ for token in doc])
plt.figure(figsize = (10,5))
plt.bar(pos_counts.keys() , pos_counts.values())
plt.xlabel("POS Tags")
plt.ylabel("Frequency")
plt.show()
```

# POS Visualize using displacy
# spacy can visialize part-of-speech information directly.

```python
text = "Python is a high level language."
doc = nlp(text)
options = {"compact" : True ,"distance" : 100}
spacy.displacy.render(doc , style = "dep" , options = options , jupyter = True)
```

# POS Tgging for a paragraph

```python
text = """
Quantum computing relies on qubits to process information.
These systems solve complex mathematical equations rapidly.
Researchers design advanced algorithms to explore deep space mysteries.
"""
doc = nlp(text)
for sentence in doc.sents :
  print("\nSentence : " , sentence.text)
  for token in sentence :
    print(
        f"{token.text:15} , {token.pos_:8} , {token.tag_:10} , {spacy.explain(token.tag_)}"
    )
```

#  Create a POS Table
# for teaching a table containing word + POS + explanation is very useful .

```python
import pandas as pd
text = "Automation streamlines repetitive tasks."
doc = nlp(text)
data = []
for token in doc :
  data.append({
      "Word" : token.text ,
      "POS" : token.pos_ ,
      "Meaning" : spacy.explain(token.pos_)
  })
  df = pd.DataFrame(data)
  print(df)
```

# Complete POS Visualization Program

```python
import spacy
from collections import Counter
import matplotlib.pyplot as plt
nlp = spacy.load("en_core_web_sm")
text = """
Cybersecurity protects vital digital infrastructure.
Experts deploy robust firewalls to block unauthorized access.
Network security is heavily prioritized in modern enterprises.
"""
doc = nlp(text)
print("WORD\t\tPOS")
for token in doc :
  print(f"{token.text:15} , {token.pos_}")

  pos_counts = Counter([token.pos_ for token in doc if not token.is_space])
  print("\nPOS Frequency:")
  print(pos_counts)

  plt.figure(figsize = (10,5))
  plt.bar(pos_counts.keys() , pos_counts.values())
  plt.xlabel("POS Tags")
  plt.ylabel("Frequency")
  plt.title("POS Tag Frequency")
  plt.show()
```
