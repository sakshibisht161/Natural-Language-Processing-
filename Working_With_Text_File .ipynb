# Create a text file and write data

```python
file = open("demo.txt", "w")

file.write("Hello, World!\n")
file.write("I am learning Python.")

file.close()

print("File created and data written successfully.")
```

# Reading a Text File in Python

```python
with open("demo.txt", "r") as file:
    content = file.read()
    print(content)
```

# Reading a File Using readline() in Python

```python
with open("demo.txt", "r") as file:
    line = file.readline()
    print(line)
```

# Append to a Text File in Python

```python
with open("demo.txt", "a") as file:
    file.write("\nI am learning Data Science.")
    file.write("\nFile handling is an important topic.")

print("New text added successfully.")
```

# Counting Words in a Text File

```python
with open("demo.txt", "r") as file:
    text = file.read()

words = text.split()
print( "Number Of word:" , len(words))
```

# Counting Characters 

```python
with open("demo.txt", "r") as file:
    text = file.read()

print("Number of Characters:" , len(text))
```

# Searching for a Word in a Text File

```python
search_word = input("Enter word a search:")

with open ("demo.txt","r") as file :
    data = file.read()

if search_word.lower() in data.lower():
    print(" Word found.")

else :
    print("Word not found.")
```
